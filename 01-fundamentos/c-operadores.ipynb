### 6) Deslocamento de Bits <<, >>

Deslocar bits é uma operação fundamental na computação.
Ela move os bits de um número para a esquerda ou para a direita, alterando seu valor de forma previsível e extremamente eficiente.

Essas operações são muito usadas em baixo nível, mas também aparecem em áreas como otimização, redes, criptografia e processamento de dados.

1. Aritmética Rápida (Multiplicar e Dividir por Potências de 2)

    | Operação | Equivalente Matemático      |
    | -------- | --------------------------- |
    | `x << n` | `x * 2ⁿ`                    |
    | `x >> n` | `x // 2ⁿ` (divisão inteira) |

```python
x = 5
print("x        =", x, "->", bits(x))
print("x << 1   =", x << 1, "->", bits(x << 1))  # 5 * 2
print("x << 3   =", x << 3, "->", bits(x << 3))  # 5 * 8
print("x >> 1   =", x >> 1, "->", bits(x >> 1))  # 5 // 2
```

2. Manipulação de Flags (Ligar/Desligar Bits Específicos)

    Imagine que cada bit representa um estado (ligado/desligado).

```python
READ  = 1 << 0  # 00000001
WRITE = 1 << 1  # 00000010
EXEC  = 1 << 2  # 00000100

permissoes = READ | WRITE
print("Permissões:", bits(permissoes))

# Verificar se tem permissão de escrita
print("Tem WRITE?", bool(permissoes & WRITE))
```

3. Compactação de Dados

    Podemos armazenar vários valores pequenos dentro de um único número.

```python
r, g, b = 120, 200, 75

rgb = (r << 16) | (g << 8) | b
print("RGB compactado:", rgb)

# Extraindo os valores
r2 = (rgb >> 16) & 0xFF
g2 = (rgb >> 8) & 0xFF
b2 = rgb & 0xFF

print("Extraído:", r2, g2, b2)
```

4. Criptografia

    Algoritmos criptográficos usam deslocamentos e rotações de bits para embaralhar dados.

```python
x = 0b10110011
rotacionado = ((x << 3) & 0xFF) | (x >> (8 - 3))
print("Original :", bits(x))
print("Rotacionado:", bits(rotacionado))
```

5. Processamento de Sinais e Protocolos

    Sensores, redes e protocolos usam bits para empacotar dados.

```python
pacote = 0b10101100

tipo    = (pacote >> 6) & 0b11
status  = (pacote >> 4) & 0b11
valor   = pacote & 0b1111

print("Tipo:", tipo)
print("Status:", status)
print("Valor:", valor)
```

## Operadores lógicos

Operadores lógicos servem para realizar operações entre tipos booleanos.

![alt text](../content/images/01-fundamentos/Logical-Operator.png "Operadores lógicos")

### Operador E (`and`)

O operador `E lógico` retorna `True` somente quando todos os operandos forem `True`.

```python
a = 0
b = 10
```

```python
# False       # True
( a > 0) and (b > 0)
```

```python
# True         True
( a == 0) and (b > 0)
```

```python
# revisar
# False       # False
( a != 0) and (b != 10)
```

### Operador OU (`or`)

O operador `OU lógico` retorna `True` quando pelo menos um dos operandos for `True`.

```python
a = 0
b = 10
```

```python
# False     # True
( a > 0) or (b > 0)
```

```python
# True        True
( a == 0) or (b > 0)
```

### Operador NÃO (`not`)

O operador `NÃO lógico` inverte o valor booleano, ou seja, quando `True`, o inverso fica `False` e vice-versa.

```python
(a == 0)
```

```python
not (a == 0)
```

## Operadores binários (ou bit a bit)

São similares a operadores lógicos, mas podem ser usados em variáveis numéricas ou em bibliotecas que veremos adiante como `Pandas` ou `Numpy`.

![alt text](../content/images/01-fundamentos/Bitwise-Operators.png "Operadores bit-a-bit")

```python
a = 17
b = 10

print('Número binário para {0} é {0:08b}'.format(a))
print('Número binário para {0} é {0:08b}'.format(b))
```

```python
print('Operação E binária: {0}. Em binário: {0:08b}'.format(a & b))
```

```python
print('Operação OU binária: {0}. Em binário: {0:08b}'.format(a | b))
```

```python
print('Operação OU Exclusivo binária: {0}. Em binário: {0:08b}'.format(a ^ b))
```

```python
print('Operação NÃO binária: {0}. Em binário: {0:08b}'.format(~a))
print('Operação NÃO binária: {0}. Em binário: {0:08b}'.format(~b))
```

```python
print('Operação SHIFT LEFT binária: {0}. Em binário: {0:08b}'.format(a >> 2))
print('Operação SHIFT LEFT binária: {0}. Em binário: {0:08b}'.format(b >> 2))
```

```python
print('Operação SHIFT RIGHT binária: {0}. Em binário: {0:08b}'.format(a << 2))
print('Operação SHIFT RIGHT binária: {0}. Em binário: {0:08b}'.format(b << 2))
```

## Operações especiais

Existem 2 operadores especiais cujo resultado é booleano, mas que nem toda linguagem de programação possui.

### Operador `is`

Esse tipo de operador é bem específico. À primeira vista, os operadores `==` e `is` parecem trabalhar de formas similares.

```python
x = 10
y = 10

print(x == y)
print(x is y)
```

O operador `==` compara se o **valor** de 2 operandos são iguais enquanto que o operador `is` compara se o **endereço de memória** dos 2 operandos são iguais.

```python
x = 10
print('endereço de memória de x', id(x))

y = 10
print('endereço de memória de y', id(y))

print('==', x == y)
print('is', x is y)
```

```python
x = 10
print('endereço de memória de x', id(x))

y = 10.0
print('endereço de memória de y', id(y))

print('==', x == y)
print('is', x is y)
```

```python
l1 = [1, 2, 3, 4] # veremos listas mais adiante
print('endereço de memória de l1', id(l1))

l2 = [1, 2, 3, 4]
print('endereço de memória de l2', id(l2))

print('==', l1 == l2)
print('is', l1 is l2)
```

### Operador `in`

O operador `in` serve basicamente para verificar se um valor está contido ou é membro de um `iterable` (que veremos mais adiante).

```python
lista = [1, 2.5, 'Python', 10]

print(10 in lista)
print('Python' in lista)
print(5 in lista)
```

```python
s = 'Python'
print('P' in s)
print('z' not in s)
```
