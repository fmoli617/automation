## Operadores binários (ou bit a bit)

Os operadores bit a bit trabalham diretamente na representação binária dos números.
Ou seja, em vez de operar no valor decimal inteiro, o Python opera bit por bit, comparando as posições dos zeros e uns.

Isso é muito usado em baixo nível, mas também aparece em redes, criptografia, gráficos, permissões de sistema e protocolos de comunicação, mas também são similares a operadores lógicos, mas podem ser usados em variáveis numéricas ou em bibliotecas que veremos adiante como `Pandas` ou `Numpy`.

![alt text](../content/images/01-fundamentos/bitwise-operators.png "Operadores bit-a-bit")

```python
# vamos executar algumas expressões para ver a precedência dos operadores, conforme a tabela acima, irei adicionar essa função auxiliar para facilitar a 
# visualização de operações relacionadas a bitts

def bits(n): return format(n & 0xFF, '08b')

def mostrar_op(a, b, op ):
    """Mostra a operação bit a bit alinhada em colunas."""
    ops = {
        "&": (lambda x, y: x & y, "AND"),
        "|": (lambda x, y: x | y, "OR"),
        "^": (lambda x, y: x ^ y, "XOR"),
    }
    f, nome = ops[op]
    r = f(a, b)
    print(f"{nome}  ({op})")
    print(f"a = {a:>4} -> {bits(a)}")
    print(f"b = {b:>4} -> {bits(b)}")
    print(f"    {op*len(bits(0))}")
    print(f"r = {r:>4} -> {bits(r)}")
    print()
```

### 1) AND &

“O AND bit a bit só mantém o 1 quando os dois lados têm 1.”

Imagine que cada número vira uma fileira de lâmpadas acesas (1) ou apagadas (0).
O operador & só deixa a lâmpada acesa se as duas estiverem acesas na mesma posição.

Exemplo:

```python
a, b = 6, 3
mostrar_op(a, b, "&")
```

### 2) OR |

“O OR bit a bit liga a luz se pelo menos um dos lados tiver luz.”

Aqui basta um dos bits ser 1 para o resultado ser 1.

Exemplo:

```python
a, b = 6, 3
mostrar_op(a, b, "|")
```

### 3) XOR ^

“O XOR liga o bit apenas quando os dois são diferentes.”

Se os bits forem iguais, vira 0. Se forem diferentes, vira 1.

```python
a, b = 6, 3
mostrar_op(a, b, "^")
```

### 4) NOT ~ (e a regra ~x = -(x + 1))

“O NOT bit a bit inverte todas as luzes.”

Mas no computador isso gera um número negativo, por causa da forma como números negativos são representados (complemento de dois).

Exemplo:

```python
x = 5
print("NOT (~)")
print(f"x   = {x:>4} -> {bits(x)}")
print(f"~x  = {~x:>4} -> {bits(~x)}")

print()

print("Regra em Python:")
print("~x == -(x + 1)  ->", ~x == -(x + 1))
print("-(x + 1) =", -(x + 1))
```

### 5) Shift à esquerda << (`x * 2ⁿ`)

Ela move os bits de um número para a esquerda, alterando seu valor de forma previsível e extremamente eficiente.

```python
x = 5
print("Shift à esquerda (<<)")
print(f"x       = {x:>4} -> {bits(x)}")
print(f"x << 1  = {x<<1:>4} -> {bits(x<<1)}   (x * 2)")
print(f"x << 2  = {x<<2:>4} -> {bits(x<<2)}   (x * 4)")
print(f"x << 3  = {x<<3:>4} -> {bits(x<<3)}   (x * 8)")
```

### 6) Célula — Shift à direita >> (`x // 2ⁿ` (divisão inteira))

Ela move os bits de um número para a direita, alterando seu valor de forma previsível e extremamente eficiente.

```python
x = 5
print("Shift à direita (>>)")
print(f"x       = {x:>4} -> {bits(x)}")
print(f"x >> 1  = {x>>1:>4} -> {bits(x>>1)}   (x // 2)")
print(f"x >> 2  = {x>>2:>4} -> {bits(x>>2)}   (x // 4)")
print(f"x >> 3  = {x>>3:>4} -> {bits(x>>3)}   (x // 8)")
```

### 7) Shift à direita com negativo (preserva o sinal)

```python
x = -5
print("Shift à direita (>>) com número negativo")
print(f"x       = {x:>4} -> {bits(x)}")
print(f"x >> 1  = {x>>1:>4} -> {bits(x>>1)}")
print(f"x >> 2  = {x>>2:>4} -> {bits(x>>2)}")
```

### 8) Exemplos

#### a. Aritmética Rápida (Multiplicar e Dividir por Potências de 2)

| Operação | Equivalente Matemático      |
| -------- | --------------------------- |
| `x << n` | `x * 2ⁿ`                    |
| `x >> n` | `x // 2ⁿ` (divisão inteira) |

```python
x = 5  # Valor inicial
print("x         =", x, "->", bits(x))  

# Cada deslocamento à esquerda multiplica por 2 (parte inteira)
# Desloca 1 bit para a esquerda
print("x << 1   =", x << 1, "->", bits(x << 1))  # 5 * (2**1) = 5 * 2 = 10

# Desloca 2 bits para a esquerda
print("x << 2   =", x << 2, "->", bits(x << 2))  # 5 * (2**2) = 5 * 4 = 40

print()

# Cada deslocamento à direita divide por 2 (parte inteira)
# Desloca 1 bit para a direita
print("x >> 1   =", x >> 1, "->", bits(x >> 1))  # 5 // (2**1) = 5 // 2 = (2.5)  = 2

print("x >> 2   =", x >> 2, "->", bits(x >> 2))  # 5 // (2**2) = 5 // 4 = (1.25) = 1
```

#### b. Manipulação de Flags (Ligar/Desligar Bits Específicos)

Imagine que cada bit representa um estado (ligado/desligado).

```python
# Cada constante representa um bit específico
READ  = 1 << 0  # 00000001 → bit 0
WRITE = 1 << 1  # 00000010 → bit 1
EXEC  = 1 << 2  # 00000100 → bit 2

# Combina permissões usando OR bit a bit
# Resultado terá bits de READ e WRITE ligados
permissoes = READ | WRITE
print("Permissões:", bits(permissoes))

# Verifica se o bit WRITE está ativado
# AND bit a bit isola o bit desejado
print("Tem WRITE?", bool(permissoes & EXEC))
```

#### c. Compactação de Dados

Podemos armazenar vários valores pequenos dentro de um único número.

Aqui estamos guardando três valores (R, G, B), cada um de 0 a 255 (8 bits), dentro de um único inteiro de 24 bits:

- a) Compactação

    `(r << 16)` empurra o vermelho 16 bits para a esquerda → ele vai para o “bloco” de cima

    `(g << 8)` empurra o verde 8 bits → ele fica no “meio”

    `b` fica nos últimos 8 bits

    depois usamos `|` para juntar tudo num único número

- b) Extração

    `rgb >> 16` traz o bloco do vermelho para o final (direita)

    `& 0xFF` pega apenas os últimos 8 bits (máscara 11111111)

    O mesmo vale para o verde com `>> 8`, e o azul é só `& 0xFF`.

```python
r, g, b = 120, 200, 75

rgb = (r << 16) | (g << 8) | b
print("RGB compactado:", rgb)

# Extraindo os valores
r2 = (rgb >> 16) & 0xFF
g2 = (rgb >> 8) & 0xFF
b2 = rgb & 0xFF

print("Extraído:", r2, g2, b2)
```

#### d. Criptografia

Algoritmos criptográficos usam deslocamentos e rotações de bits para embaralhar dados.

Um deslocamento normal perde bits: ao deslocar para a esquerda, bits “caem fora”.
Na rotação, os bits que “saem” de um lado voltam pelo outro.

Aqui fazemos uma rotação de 8 bits para a esquerda em 3 posições.

Explicando as duas partes:

-  a) Parte esquerda:

    `(x << 3)` desloca 3 para a esquerda

    `& 0xFF` garante que fiquemos só com 8 bits (remove o “excesso”)

- b) Parte direita:

    `x >> (8 - 3)` pega exatamente os bits que “caíram fora” e coloca no final

    Junta tudo:

    `|` combina as duas metades formando a rotação

```python
x = 0b10110011
rotacionado = ((x << 3) & 0xFF) | (x >> (8 - 3))
print("Original :", bits(x))
print("Rotacionado:", bits(rotacionado))
```

#### e. Processamento de Sinais e Protocolos

Sensores, redes e protocolos usam bits para empacotar dados.

- a) Imagine que pacote é um byte (8 bits) contendo 3 informações:

    `tipo`: 2 bits mais à esquerda (bits 6–7)
    `status`: 2 bits seguintes (bits 4–5)
    `valor`: 4 bits finais (bits 0–3)

- b) Extraindo:

    `>> 6` traz os 2 bits mais altos para o final

    `& 0b11` pega só 2 bits (11)

    `> 4` traz o bloco do status para o final

    `& 0b11` pega só 2 bits

    `& 0b1111` pega só os 4 últimos bits

```python
pacote = 0b10101100

tipo    = (pacote >> 6) & 0b11
status  = (pacote >> 4) & 0b11
valor   = pacote & 0b1111

print("Tipo:", tipo)
print("Status:", status)
print("Valor:", valor)
```

## Operações especiais

Existem 2 operadores especiais cujo resultado é booleano, mas que nem toda linguagem de programação possui.

### 1) Operador `is`

Esse tipo de operador é bem específico. À primeira vista, os operadores `==` e `is` parecem trabalhar de formas similares.

```python
x = 10
y = 10

print(x == y)
print(x is y)
```

```python
x = 10
print('endereço de memória de x', id(x))

y = 10
print('endereço de memória de y', id(y))

print('==', x == y)
print('is', x is y)
```

```python
s1 = "python"
s2 = "python"

print(s1 is s2)  # Geralmente True
```

```python
a = [1, 2, 3]
b = a

print(a is b)  # True
print(id(a), id(b))
```

```python
x = 10
print('endereço de memória de x', id(x))

y = 10.0
print('endereço de memória de y', id(y))

print('==', x == y)
print('is', x is y)
```

```python
l1 = [1, 2, 3, 4] # veremos listas mais adiante
print('endereço de memória de l1', id(l1))

l2 = [1, 2, 3, 4]
print('endereço de memória de l2', id(l2))

print('==', l1 == l2)
print('is', l1 is l2)
```

### 2) Operador `in`

O operador `in` serve basicamente para verificar se um valor está contido ou é membro de um `iterable` (que veremos mais adiante).

```python
lista = [1, 2.5, 'Python', 10]

print(10 in lista)
print('Python' in lista)
print(5 in lista)
```

```python
s = 'Python'
print('P' in s)
print('z' not in s)
```
