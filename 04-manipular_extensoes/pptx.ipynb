```python
import os
import win32com.client


def pptx_para_imagens(caminho_pptx: str, pasta_saida: str) -> None:
    os.makedirs(pasta_saida, exist_ok=True)

    powerpoint = win32com.client.Dispatch("PowerPoint.Application")
    powerpoint.Visible = 1  # Deixa visível para fins didáticos

    presentation = powerpoint.Presentations.Open(os.path.abspath(caminho_pptx))

    # Exporta todos os slides como PNG
    presentation.Export(os.path.abspath(pasta_saida), "PNG")

    presentation.Close()
    powerpoint.Quit()


if __name__ == "__main__":
    pptx_para_imagens(
        caminho_pptx=r"C:\projetos\automation\input\pwc_finance_function_business_partnering.pptx",
        pasta_saida=r"output"
    )
    print("Conversão concluída!")
```

```python
from pathlib import Path


def contar_pngs(pasta_saida: str) -> int:
    pasta = Path(pasta_saida)
    return len(list(pasta.glob("*.png"))) + len(list(pasta.glob("*.PNG")))


def validar_saida(pasta_saida: str) -> None:
    pasta = Path(pasta_saida)

    assert pasta.exists(), "Pasta de saída não existe"
    assert pasta.is_dir(), "Saída não é uma pasta"

    total = contar_pngs(pasta_saida)
    assert total > 0, "Nenhuma imagem PNG foi gerada"

    slide1_png = pasta / "Slide1.PNG"
    slide1_png2 = pasta / "Slide1.png"
    assert slide1_png.exists() or slide1_png2.exists(), "Slide1 não encontrado"

    print(f"✅ Teste OK — {total} imagens encontradas!")


if __name__ == "__main__":
    validar_saida("output")

```
